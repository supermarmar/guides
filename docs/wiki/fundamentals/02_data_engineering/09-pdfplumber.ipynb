## **Text**

Extract Text `pdfplumber` provides several methods for extracting text from a PDF. The simplest method is the `extract_text()` method, which returns a string containing all the text in the PDF. For example:

```python
with pdfplumber.open("basel_framework.pdf") as pdf:
    first_page = pdf.pages[316]
    text = first_page.extract_text()
    print(text)
```

## **Tables**

Extract Data `pdfplumber` also provides several methods for extracting data from a PDF. One such method is the `extract_table()` method, which returns a list of lists containing the data from tables in the PDF. For example:

```python
with pdfplumber.open("basel_framework.pdf") as pdf:
    first_page = pdf.pages[318]
    table = first_page.extract_tables()
    df = pd.DataFrame(table[0][1:], columns=table[0][0])

df.head()
```
